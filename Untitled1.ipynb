```python
import pandas as pd
import numpy as np
from sklearn.base import BaseEstimator # Added BaseEstimator
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.svm import OneClassSVM
from sklearn.ensemble import IsolationForest, VotingClassifier, StackingClassifier, RandomForestClassifier # Added RandomForestClassifier
from sklearn.linear_model import LogisticRegression

# Define column names for the NSL-KDD dataset
# These are standard columns for NSL-KDD from UCI Machine Learning Repository
colnames = [
    'duration', 'protocol_type', 'service', 'flag', 'src_bytes', 'dst_bytes', 'land', 'wrong_fragment', 'urgent', 'hot',
    'num_failed_logins', 'logged_in', 'num_compromised', 'root_shell', 'su_attempted', 'num_root', 'num_file_creations',
    'num_shells', 'num_access_files', 'num_outbound_cmds', 'is_host_login', 'is_guest_login', 'count', 'srv_count',
    'serror_rate', 'srv_serror_rate', 'rerror_rate', 'srv_rerror_rate', 'same_srv_rate', 'diff_srv_rate',
    'srv_diff_host_rate', 'dst_host_count', 'dst_host_srv_count', 'dst_host_same_srv_rate',
    'dst_host_diff_srv_rate', 'dst_host_same_src_port_rate', 'dst_host_srv_diff_host_rate', 'dst_host_serror_rate',
    'dst_host_srv_serror_rate', 'dst_host_rerror_rate', 'dst_host_srv_rerror_rate', 'attack_type', 'difficulty_level'
]

# --- Load the NSL-KDD Dataset ---
# Please make sure 'KDDTrain+.txt' and 'KDDTest+.txt' are available in the current directory or provide their full paths.
# Example: If they are in a 'data' folder: 'data/KDDTrain+.txt'
try:
    df_train = pd.read_csv('KDDTrain+.txt', names=colnames)
    df_test = pd.read_csv('KDDTest+.txt', names=colnames)
    print("Datasets loaded successfully")
except FileNotFoundError:
    print("Error: KDDTrain+.txt or KDDTest+.txt not found. Please ensure the files are in the correct directory.")
    # As a fallback for demonstration if files are not present, create dummy data
    print("Generating dummy data for demonstration purposes.")
    dummy_data = {col: np.random.rand(100) if col not in ['protocol_type', 'service', 'flag', 'attack_type']
                  else np.random.choice(['tcp', 'udp', 'icmp'] if col == 'protocol_type'
                                      else ['http', 'ftp'] if col == 'service'
                                      else ['SF', 'S0'] if col == 'flag'
                                      else ['normal', 'attack'], 100)
                  for col in colnames}
    df_train = pd.DataFrame(dummy_data)
    df_test = pd.DataFrame(dummy_data)

# Combine for consistent preprocessing, then separate later
df_combined = pd.concat([df_train.drop(columns=['difficulty_level']), df_test.drop(columns=['difficulty_level'])], ignore_index=True)

print("Combined data shape:", df_combined.shape)
```

```python
# --- Preprocessing ---

# Identify categorical and numerical features
categorical_features = [
    'protocol_type', 'service', 'flag'
]

numerical_features = [
    'duration', 'src_bytes', 'dst_bytes', 'land', 'wrong_fragment', 'urgent', 'hot',
    'num_failed_logins', 'logged_in', 'num_compromised', 'root_shell', 'su_attempted', 'num_root',
    'num_file_creations', 'num_shells', 'num_access_files', 'num_outbound_cmds', 'is_host_login',
    'is_guest_login', 'count', 'srv_count', 'serror_rate', 'srv_serror_rate', 'rerror_rate',
    'srv_rerror_rate', 'same_srv_rate', 'diff_srv_rate', 'srv_diff_host_rate', 'dst_host_count',
    'dst_host_srv_count', 'dst_host_same_srv_rate', 'dst_host_diff_srv_rate',
    'dst_host_same_src_port_rate', 'dst_host_srv_diff_host_rate', 'dst_host_serror_rate',
    'dst_host_srv_serror_rate', 'dst_host_rerror_rate', 'dst_host_srv_rerror_rate'
]

# Apply One-Hot Encoding to categorical features
df_combined_encoded = pd.get_dummies(df_combined, columns=categorical_features, drop_first=True)

# Separate features (X) and target (y)
X = df_combined_encoded.drop(columns=['attack_type'])
y = df_combined_encoded['attack_type']

# Map attack types to binary labels (0 for normal, 1 for attack)
y_binary = y.apply(lambda x: 0 if x == 'normal' else 1)

# Scale numerical features
scaler = StandardScaler()
X[numerical_features] = scaler.fit_transform(X[numerical_features])

print("Preprocessing complete. Feature shape:", X.shape, "Target shape:", y_binary.shape)
```

```python
# --- Prepare Training and Test Sets ---

# Re-split the preprocessed data back into train and test based on original dataset sizes
# The original df_train did not have 'difficulty_level' when concatenated
X_train_full = X.iloc[:len(df_train)]
y_train_full = y_binary.iloc[:len(df_train)]

X_test_full = X.iloc[len(df_train):]
y_test = y_binary.iloc[len(df_train):]

# For unsupervised anomaly detection, we train only on 'normal' samples from the training set.
# Filter X_train_full to include only normal samples
X_train_normal = X_train_full[y_train_full == 0]

# For the test set, we use the full X_test_full and y_test_full as X_test and y_test
X_test = X_test_full

print("X_train_normal shape (for anomaly detection training):", X_train_normal.shape)
print("X_test shape (for evaluation):", X_test.shape)
print("y_test shape (for evaluation):", y_test.shape)

# Ensure y_test is a numpy array for consistent use with sklearn metrics
y_test = y_test.values
```

```python
print("\nAnomaly Detection Models")

from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import MinMaxScaler
import inspect # Used to inspect init signature for dynamic parameter handling
from sklearn.svm import OneClassSVM # Ensure these are imported if not already
from sklearn.ensemble import IsolationForest # Ensure these are imported if not already

# --- New: AnomalyScoreOnlyTransformer for use in Pipelines ---
class AnomalyScoreOnlyTransformer(BaseEstimator, TransformerMixin):
    def __init__(self, model_class, nu=None, kernel=None, gamma='auto', contamination=None, random_state=None):
        self.model_class = model_class
        self.nu = nu
        self.kernel = kernel
        self.gamma = gamma
        self.contamination = contamination
        self.random_state = random_state
        self.model = None
        self.scaler = MinMaxScaler() # To scale anomaly scores consistently

    def _get_model_params(self):
        # Filter parameters relevant to the underlying model_class from self.get_params()
        model_init_signature = inspect.signature(self.model_class.__init__)
        actual_model_params = {}
        wrapper_params = self.get_params() # This gets all params from AnomalyScoreOnlyTransformer's __init__
        for param_name, param_obj in model_init_signature.parameters.items():
            if param_name != 'self' and param_name in wrapper_params and wrapper_params[param_name] is not None:
                actual_model_params[param_name] = wrapper_params[param_name]
        return actual_model_params

    def fit(self, X, y=None):
        # Anomaly detectors should ideally be fitted on 'normal' data only.
        # If 'y' is provided, filter X to include only normal samples (y=0).
        X_normal = X[y == 0] if y is not None else X

        self.model = self.model_class(**self._get_model_params())
        self.model.fit(X_normal)

        # Fit a scaler on the anomaly scores generated from the normal training data (X_normal).
        # This helps normalize scores before passing to the next stage in the pipeline.
        scores_normal = self._get_raw_scores(X_normal)
        # Reshape for scaler (expects 2D array)
        self.scaler.fit(scores_normal.reshape(-1, 1))

        return self

    def _get_raw_scores(self, X):
        print(f"DEBUG: Inside _get_raw_scores. self.model type: {type(self.model)}")
        print(f"DEBUG: hasattr(self.model, 'decision_function'): {hasattr(self.model, 'decision_function')}")
        print(f"DEBUG: hasattr(self.model, 'score_samples'): {hasattr(self.model, 'score_samples')}")

        if hasattr(self.model, 'decision_function'):
            scores = self.model.decision_function(X)
        elif hasattr(self.model, 'score_samples'):
            scores = self.model.score_samples(X)
        else:
            print(f"DEBUG: Error raised because model {type(self.model)} lacks decision_function/score_samples.")
            raise AttributeError("Underlying model must have 'decision_function' or 'score_samples'.")
        return scores

    def transform(self, X):
        raw_scores = self._get_raw_scores(X)

        # Anomaly detector scores typically have higher values for 'normal' instances.
        # We want scores where higher values mean 'anomaly' for a binary classifier.
        # So, we invert the scores (e.g., score -> -score).
        inverted_scores = -raw_scores

        # Scale scores to a [0, 1] range using the scaler fitted on normal data.
        # This makes them interpretable as 'anomaly probability' for the meta-classifier.
        scaled_scores = self.scaler.transform(inverted_scores.reshape(-1, 1))
        return scaled_scores

# Helper function to create a pipeline for anomaly detection followed by classification
def create_anomaly_pipeline(model_class, **model_params):
    pipeline_steps = [
        ('anomaly_score_transformer', AnomalyScoreOnlyTransformer(model_class, **model_params)),
        # The meta-classifier learns to classify based on the single anomaly score feature
        ('meta_classifier', LogisticRegression(solver='liblinear', random_state=42))
    ]
    return Pipeline(pipeline_steps)

# --- Instantiate Anomaly Detection Pipelines ---
# One-Class SVM Pipeline
oc_svm_pipeline = create_anomaly_pipeline(OneClassSVM, nu=0.1, kernel="rbf", gamma='auto')

# Isolation Forest Pipeline
iso_forest_pipeline = create_anomaly_pipeline(IsolationForest, contamination=0.1, random_state=42)

print("Anomaly detection pipelines instantiated for One-Class SVM and Isolation Forest.")
```

```python
# -------------------------- Ensemble Methods --------------------------
print("\n🔄 Training Ensemble Models...")

# Use the newly created pipelines as estimators
estimators = [
    ('oc_svm_pipeline', oc_svm_pipeline),
    ('iso_forest_pipeline', iso_forest_pipeline)
]

# 1. Voting
# Use soft voting as the pipelines output probabilities via LogisticRegression.
# The pipelines will now be trained on X_train_full and y_train_full.
voting = VotingClassifier(estimators=estimators, voting='soft')
voting.fit(X_train_full, y_train_full)

# 2. Blending
# Split X_train_full further to create a holdout set for the meta-learner.
# The base models (pipelines) will be fitted on X_main, and their predictions
# on X_hold will form the meta-features for the meta-learner.
X_main, X_hold, y_main, y_hold = train_test_split(X_train_full, y_train_full, test_size=0.2, random_state=42)

# Re-instantiate pipelines for blending to avoid data leakage
oc_svm_blend_pipeline = create_anomaly_pipeline(OneClassSVM, nu=0.1, kernel="rbf", gamma='auto')
oc_svm_blend_pipeline.fit(X_main, y_main)

iso_blend_pipeline = create_anomaly_pipeline(IsolationForest, contamination=0.1, random_state=42)
iso_blend_pipeline.fit(X_main, y_main)

# Get predictions (probabilities of class 1, i.e., anomaly) from base pipelines on the holdout set
meta_features_blend = np.column_stack([
    oc_svm_blend_pipeline.predict_proba(X_hold)[:, 1],
    iso_blend_pipeline.predict_proba(X_hold)[:, 1]
])

# Train a meta-learner (Logistic Regression) on these meta-features
meta_learner = LogisticRegression(solver='liblinear', random_state=42).fit(meta_features_blend, y_hold)

# Generate meta-features for the test set (using probabilities of being an attack)
# IMPORTANT: Use the _blend_pipeline instances, as they were the ones fitted in this blending context.
test_meta = np.column_stack([
    oc_svm_blend_pipeline.predict_proba(X_test)[:, 1],
    iso_blend_pipeline.predict_proba(X_test)[:, 1]
])

# Make final predictions using the meta-learner
y_pred_blend = meta_learner.predict(test_meta)

# 3. Stacking
# StackingClassifier also benefits from 'predict_proba' (from the pipelines).
stacking = StackingClassifier(estimators=estimators, final_estimator=RandomForestClassifier(random_state=42, class_weight='balanced', n_estimators=200, max_depth=10), cv=5)
stacking.fit(X_train_full, y_train_full)

# Final Predictions on the test set
y_pred_vote = voting.predict(X_test)
y_pred_stack = stacking.predict(X_test)

print("Ensemble models trained and predictions generated using pipelines.")
```
